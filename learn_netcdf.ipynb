# 🌐 NetCDF (Network Common Data Form) Reference Guide

## 1. Overview
* **Full Name:** Network Common Data Form
* **Developer:** Maintained by **Unidata** (a program of the University Corporation for Atmospheric Research - UCAR).
* **Core Purpose:** An open-standard, self-describing, machine-independent binary format designed to store and manage multi-dimensional scientific arrays.
* **Primary Domains:** Meteorology 🌤️, Oceanography 🌊, Climate Modeling 🌡️, and Atmospheric Sciences.

---

## 2. File Extensions & Variants

| Extension | Variant | Details |
| :--- | :--- | :--- |
| `.nc` / `.cdf` | **NetCDF Classic (NetCDF-3)** | Legacy format; limited file sizes (< 2GB/4GB) and simpler structure. |
| `.nc` / `.nc4` | **NetCDF-4** | Current standard; uses **HDF5** as its underlying storage engine, enabling compression, parallel I/O, and massive file sizes. |

---

## 3. The NetCDF Data Model

NetCDF organizes data around three core components:

### 📐 Dimensions
Define the physical axes or grid parameters of the dataset.
* **Fixed Dimensions:** Have a defined length (e.g., `lat = 180`, `lon = 360`).
* **Unlimited Dimensions:** Can grow infinitely along an axis (e.g., `time` growing as real-time observational data streams in).

### 📊 Variables
Multi-dimensional arrays of values belonging to a single data type (float, integer, character).
* Tied directly to defined dimensions.
* *Example:* `temperature(time, lat, lon)` maps temperature readings across every 3D coordinate point.

### 🏷️ Attributes
Metadata that provides context to the raw data.
* **Global Attributes:** Describe the dataset as a whole (e.g., `institution = "NOAA"`, `title = "Global Sea Surface Temperature"`).
* **Variable Attributes:** Describe specific variables (e.g., `units = "Kelvin"`, `valid_range = [200, 350]`, `_FillValue = -9999`).

---

## 4. Key Python Ecosystem

* **`netCDF4`:** Low-level C-library wrapper for direct creation, writing, and byte manipulation.
* **`xarray`:** High-level library built on top of NumPy and Pandas; the gold standard for reading and analyzing NetCDF files cleanly using labeled coordinates.

```python
import netCDF4 as nc
import xarray as xr
import numpy as np

# Create a ncdf4 file in write mode
with nc.Dataset("my_first_dataset.nc", "w", format="NETCDF4") as ds:
  # Define Dimensions
  ds.createDimension("time", 3)
  ds.createDimension("lat", 2)
  ds.createDimension("lon", 2)

  # Create a 3d array
  temp = ds.createVariable("temperature", "f4", ("time", "lat", "lon"))

  # Create Attributes
  ds.title = "Synthetic Temperature Data"
  temp.units = "Celsius"

  # Fill Data
  temp[:] = np.array([
      [[20.1, 21.3], [19.8, 20.5]],
      [[20.1, 21.3], [19.8, 20.5]],
      [[20.1, 21.3], [19.8, 20.5]],
  ])
  print("Sucess")
```
